```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
spotify_df = pd.read_csv('../data/Cleaned_Spotify_2024_Global_Streaming_Data.csv')
spotify_df.head()
```

```python
#group by genre and calculate the average total streams
avg_streams = spotify_df.groupby('Genre')['Total Streams (Millions)'].mean().sort_values(ascending=False)
avg_streams
```

```python
plt.bar(avg_streams.index, avg_streams.values)
plt.show()
```

```python
plt.bar(avg_streams.index, avg_streams.values)

plt.xlabel('Genre')
plt.ylabel('Average Total Streams (Millions)')
plt.title('Average Total Streams by Genre')

plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(avg_streams.index, avg_streams.values)

plt.xlabel('Genre')
plt.ylabel('Average Total Streams (Millions)')
plt.title('Average Total Streams by Genre')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.bar(avg_streams.index, avg_streams.values)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel('Genre')
plt.ylabel('Average Total Streams (Millions)')
plt.title('Average Total Streams by Genre')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
avg_artist_streams = spotify_df.groupby("Artist")["Total Streams (Millions)"].mean().sort_values(ascending=False)
avg_artist_streams
```

```python
plt.Figure(figsize=(10,6))
colors = ['#31a354']*len(avg_artist_streams)

max_index = avg_artist_streams.idxmax()
highlight_position = avg_artist_streams.index.get_loc(max_index)
colors[highlight_position] = '#f030ad'

plt.bar(avg_artist_streams.index, avg_artist_streams.values, color = colors, width=0.8)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel('Artist')
plt.ylabel('Average Total Streams (Millions)')
plt.title('Average Total Streams by Artist')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12,8))

special_artist = "Ariana Grande"
colors = ["#f064bf" if artist == special_artist else "gray" for artist in avg_artist_streams.index] #created a for loop to iterate through avg_artist_streams and it's going to check each artist to see if it's just artist or special_artist

plt.bar(avg_artist_streams.index, avg_artist_streams.values, color = colors, width=0.8)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel('Artist')
plt.ylabel('Average Total Streams (Millions)')
plt.title('Ariana Grande Ranks 2nd in Total Average Streams')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

featured_genre = 'Pop'

colors = ['#08cc15' if genre == featured_genre else '#d1d5d1' for genre in spotify_df['Genre']]

plt.scatter(
    spotify_df['Streams Last 30 Days (Millions)'],
    spotify_df['Total Streams (Millions)'],
    color = colors,
    alpha = 0.8,
    s = 25)

plt.xlabel('Streams this month')
plt.ylabel('Total Streams') 

ax = plt.gca() #stands for get current axes
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.set_title('Do Pop Songs Achieve Higher Monthly Listens or Total Streams?', fontsize = 18, pad = 35)

for spine in ax.spines.values():
    spine.set_linewidth(0.25)
    spine.set_alpha(0.5)

plt.tight_layout()
plt.show()
```

```python
yearly_avg = spotify_df.groupby('Release Year')['Total Streams (Millions)'].mean().round(2)
yearly_avg
```

```python
plt.figure(figsize=(12,6))
plt.plot(yearly_avg.index, 
         yearly_avg.values,
         marker='.',
         linestyle = '-',
         color = "#3fbd8f",
         linewidth = 1.5)

plt.xlabel("Release Year")
plt.ylabel("Average Total Streams")
plt.title("Did Listeners Stream Less in 2022?", fontsize = 18, pad = 20)

ax = plt.gca() #get current axes
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.axvline(  
    2022,           #puts a vertical line at the year 2022
    color = "#e8622e",
    linewidth = 1,
    alpha = 0.7)        #alpha controls the transparency of an item        

ax.text(
    2022+0.05,
    2225,
    "Dip In Total Streams",
    color = "#e8622e",  #You can create a name/variable for this color at the beginning to save time and for consistency in colors
    fontsize = 8       
)

plt.ylim(0, 3500)

plt.tight_layout()         #Adjusts the padding between and around subplots; looks like it makes everything bigger, covers more area
plt.show()
```

```python
#Pie charts - a bit controversial; useful for 2-3 categories, not useful for more categories
category = "Platform Type"
values = "Total Streams (Millions)"

platform_totals = spotify_df.groupby(category)[values].sum()    #to get the sum of the values by category
platform_totals
```

```python
plt.figure(figsize=(8,8))         #length, width
plt.pie(platform_totals,
            labels = None,
            startangle=110,     #changes the angle of the pie chart
            colors = ["#f8b2a5", "#c51f1f"],
            autopct = "%1.1f%%")        #put your colors in a list

plt.title("Near Even Split Between Premium and Free Users")

plt.legend(
    labels = platform_totals.index,
    loc="best",
    title = "Platform" 
)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,8))

wedges, texts, autotexts = ax.pie(
    platform_totals,
    autopct = "%1.1f%%",
    startangle=110,     #changes the angle of the pie chart
    colors = ["#f8b2a5", "#c51f1f"],
    )        #put your colors in a list

ax.set_title(
    "Near Even Split Between Premium and Free Users",
    fontsize =18,
    pad = 15,
    loc = "center")

for autotext in autotexts:
    autotext.set_fontsize(18)
    autotext.set_color("white")
    autotext.set_weight("bold")

ax.legend(
    wedges,
    platform_totals.index,
    loc="lower center",
    bbox_to_anchor = (0.5, 0.95),
    ncol = 2
)

plt.show()
```
